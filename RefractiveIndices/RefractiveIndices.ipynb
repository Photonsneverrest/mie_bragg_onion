Poly(2-Vinylpyridine)
using Cauchy model n(lambda) = A + B/lambda^2+C/lambda^4
Source:     A and B parameter for Mw = 650 and kg/mol from https://doi.org/10.1063/5.0062054
            C from https://doi.org/10.1063/1.4953855
Extinction coefficient is assumed to be zero

```python
wl_P2VP = np.arange(start_visible,end_visible+1,1)/mToN # in micrometer
A_P2VP = 1.5543
B_P2VP = 0.0101     #um^2
C_P2VP = 0.00038    #um^4
n_P2VP = A_P2VP+B_P2VP/(wl_P2VP**2)+C_P2VP/(wl_P2VP**4)
# print(wl_P2VP[np.where(wl_P2VP==0.589)]*mToN)
# print(n_P2VP[np.where(wl_P2VP==0.589)])
```

Beta carotin

Enhancing the Refractive Index of Polymers with
a Plant-Based Pigment
Mohammad Yasir, Tianqi Sai, Alba Sicher, Frank Scheffold, Ullrich Steiner,
Bodo D. Wilts, and Eric R. Dufresne

```python
betaCn = pd.read_csv(file_path + 'BC_n.csv', header=None)
wl_n_betaC = betaCn[0].to_numpy() # [nm]
n_betaC = betaCn[1].values
betaCk = pd.read_csv(file_path + 'BC_k.csv', header=None)
wl_k_betaC = betaCk[0].to_numpy() # [nm]
k_betaC_mol = betaCk[1].values # k is molar extinction coefficient [M^-1*cm^-1]

# Molar concentration of 0.1mM BC in chloroform
conc_BC = 1e-4 # M
# Extinction coefficient
k_betaC = k_betaC_mol*1e2*conc_BC*wl_k_betaC*1e-9/(4*np.pi*np.log10(np.e))

# Interpolation to have the same wavelength arrays
wl_betaC = wl_n_betaC # [nm]
k_betaC = np.interp(wl_betaC, wl_k_betaC, k_betaC)
```

```python
# Plot of refractive indices

# Create the figure and axes
fig, ax1 = plt.subplots()

surroundingColorReal = 'tab:olive'
alphaSurrounding = 0.5
linewidthSurrounding = 6

# Plot refractive indices
ax1.plot(wl_TiO2_rut_eo*mToN, n_TiO2_rut_eo, color= 'b', linestyle=':', label='$TiO_2 \enspace rut \enspace eo$')
# ax1.plot(wl_TiO2_rut_eo_2, n_TiO2_rut_eo_2, color= 'b', alpha=0.5, linestyle=':', label='$TiO_2 \enspace rut \enspace eo$')
ax1.plot(wl_TiO2_rut_o*mToN, n_TiO2_rut_o, color= 'b', linestyle='--', label='$TiO_2 \enspace rut \enspace o$')
ax1.plot(wl_TiO2*mToN, n_TiO2, color= 'b', label='$TiO_2 \enspace an$')
ax1.plot(wl_ZrO2*mToN, n_ZrO2, color= 'g', label='$ZrO_2$')
ax1.plot(wl_ZnO*mToN, n_ZnO, color= 'tab:purple', label='$ZnO$')
# ax1.plot(wl_PS*mToN, n_PS, color= '#66ff33', label='$PS$')
for i in np.arange(x_hPS.size):
    ax1.plot(wl_PS_x[i,:], n_PS_x[i,:], color= 'tab:orange', alpha = 1-(i)/(x_hPS.size+1), linewidth=0.8, label='$PS \enspace x_{hPS}=0.$'+str(x_hPS[i]))
    break
# ax1.plot(wl_P2VP*mToN, n_P2VP, color= '#FFD700', label='$P2VP$')
for i in np.arange(x_PDP.size):
    ax1.plot(wl_P2VP_x[i,:], n_P2VP_x[i,:], color= 'k', alpha = 1-(i)/(x_PDP.size+1), linewidth=0.8, label='$P2VP \enspace x_{PDP}=0.$'+str(x_PDP[i]))
    break
ax1.plot(wl_PMMA*mToN, n_PMMA, color= 'tab:brown', label='$PMMA$')
ax1.plot(wl_betaC, n_betaC, color= 'tab:pink', label='$beta - carotene$')

# Add labels and title
ax1.set_xlabel('$\lambda$ [nm]')
ax1.set_ylabel('$\Re (n)$ [-]')
ax1.xaxis.grid(False)
ax1.yaxis.grid(True, color='k', linestyle='--', linewidth=1.)
ax1.set_title("Nanoparticles, polymers and\npigments refractive indices")
# Set x ticks on both the bottom and top of the graph
ax1.xaxis.set_ticks_position('both')

start_x = 200
end_x = 800
# Specify the background color and alpha (transparency)
background_color = 'lightgray'
alpha = 0.8
# Add a colored background span
ax1.axvspan(start_x, start_visible, facecolor=background_color, alpha=alpha)
ax1.axvspan(end_visible, end_x, facecolor=background_color, alpha=alpha)

plt.xlim(start_x, end_x)
plt.ylim(1.,3.5)

# Add a legend
ax1.legend(loc='center left', bbox_to_anchor=(1.2, 0.5))

# Create the second set of axes for extinction coefficients
ax2 = ax1.twinx()
# Plot extinction coefficients surrounding line color
ax2.plot(wl_TiO2_rut_eo_2, k_TiO2_rut_eo_2, color=surroundingColorReal, linewidth=linewidthSurrounding, alpha=alphaSurrounding, label='')
ax2.plot(wl_TiO2_an_TF*mToN, k_TiO2_an_TF, color=surroundingColorReal, linewidth=linewidthSurrounding, alpha=alphaSurrounding, label='')
# ax2.plot(wl_PS*mToN, k_PS, color=surroundingColorReal, linewidth=linewidthSurrounding, alpha=alphaSurrounding, label='')
ax2.plot(wl_betaC, k_betaC, color=surroundingColorReal, linewidth=linewidthSurrounding, alpha=alphaSurrounding, label='')
# Extinction coefficients with alpha 0.5 but same colors as refractive index
ax2.plot(wl_TiO2_rut_eo_2, k_TiO2_rut_eo_2, color= 'b', alpha=1., linestyle=':', label='$TiO_2 \enspace rut \enspace eo$')
ax2.plot(wl_TiO2_an_TF*mToN, k_TiO2_an_TF, color= 'b', alpha=1., linestyle='-', label='$TiO_2 \enspace an$')
# ax2.plot(wl_PS*mToN, k_PS, color= 'tab:pink', alpha=1., label='$PS$')
ax2.plot(wl_betaC, k_betaC, color= 'tab:pink', alpha=1., linestyle='-', label='$beta - carotene$') # extinction coefficient

ax2.set_ylabel('$\Im (n)$ [-]', color=surroundingColorReal, alpha = 1.)
# ax2.set_ylim(0, 4)
ax2.xaxis.grid(False)
ax2.yaxis.grid(True, color='w', linestyle='--', linewidth=1.2)
# Set the y-axis scale to 'log'
ax2.set_yscale('log')

# Show the plot
plt.savefig(file_path + 'RefractiveIndices.png', dpi = 300, bbox_inches='tight')
plt.show()
```
